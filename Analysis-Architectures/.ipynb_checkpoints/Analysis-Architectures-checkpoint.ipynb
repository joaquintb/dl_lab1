```python
def parseSummary(capturedSummary):

    typeAllLayers              = []
    nParamAllLayers            = []
    nParamConvolutionalLayers  = []
    nParamFullyConnectedLayers = []
    nParamBatchNormLayers      = []

    s0 = capturedSummary.stdout.split('\n')
    layersInfo = False
    for s1 in s0:
        #print('*',s1)
    
        if s1.startswith('=========='):
            layersInfo = not layersInfo
            continue
    
        if layersInfo:
            s2 = s1.split()
            #print('**',s2)
            typeLayer = s2[0]
            numParam  = int(s2[-1].replace(',',''))
            #
            if   typeLayer.startswith('Conv2d-'):
                typeAllLayers.append('CV')
                nParamAllLayers.append(numParam)
                nParamConvolutionalLayers.append(numParam)
            elif typeLayer.startswith('Linear-'):
                typeAllLayers.append('FC')
                nParamAllLayers.append(numParam)
                nParamFullyConnectedLayers.append(numParam)
            elif typeLayer.startswith('BatchNorm2d-'):
                typeAllLayers.append('FC')
                nParamAllLayers.append(numParam)
                nParamBatchNormLayers.append(numParam)                
            elif typeLayer.startswith('AvgPool2d-') or typeLayer.startswith('MaxPool2d-') or \
                 typeLayer.startswith('AdaptiveAvgPool2d') or typeLayer.startswith('Flatten-') or \
                 typeLayer.startswith('Dropout-') or \
                 typeLayer.startswith('Tanh-') or typeLayer.startswith('ReLU-') or \
                 typeLayer.startswith('BasicConv2d-') or typeLayer.startswith('Inception') or \
                 typeLayer.startswith('BasicBlock-') or typeLayer.startswith('Bottleneck-') or \
                 typeLayer.startswith('Fire-'):
                dummy = None # Nothing to do
            else:
                print('typeLayer unknown: {}'.format(typeLayer))

    return (typeAllLayers, nParamAllLayers, \
            nParamConvolutionalLayers, nParamFullyConnectedLayers, nParamBatchNormLayers)
```

```python
def plotParamsSummary(typeAllLayers, nParamAllLayers, minPctPlot):

    tot = sum(nParamAllLayers)

    # Selects the numbers to plot
    if minPctPlot is None:
        typePlot    = typeAllLayers
        nParamsPlot = [nParamAllLayers[i]/tot for i in range(len(nParamAllLayers))]
    else:
        typePlot    = []
        nParamsPlot = []
        for i in range(len(nParamAllLayers)):
            pct = nParamAllLayers[i]/tot
            if pct > minPctPlot:
                typePlot.append(typeAllLayers[i])
                nParamsPlot.append(nParamAllLayers[i]/tot)
    nps = len(nParamsPlot)

    #print(nParamsPlot,tot)
    #print([nParamsPlot[i]/tot for i in range(len(nParamsPlot))])
    
    fig = plt.figure(figsize=(12, 9), dpi=80)

    X_axis = np.arange(nps)
    width  = 1/nps
    plt.bar(0, 0, width, color='tomato', label='FC')            ### Only for labels
    plt.bar(0, 0, width, color='mediumseagreen', label='¬FC')   ### Only for labels
    for i in range(nps):
        if typePlot[i] == 'FC':
            plt.bar(i, nParamsPlot[i], width, color='tomato')
        else:
            plt.bar(i, nParamsPlot[i], width, color='mediumseagreen')

    plt.xticks(X_axis)
    plt.xlabel('Number of layer with parameters')
    plt.ylabel('Percentage of parameters')
    plt.legend(loc="upper left")
    plt.show()
```

```python
def parseSummaryAndPlot(capturedSummary, minPctPlot=None):
    #
    def printStats(strStat,nParamStat,nParamTot):
        sumParamStat = sum(nParamStat)
        print('Number of parameters in {} Layers: {:9d} ({:5.2f}%)'.format(strStat,sumParamStat,100*sumParamStat/nParamTot))

    #
    ### Parse summary
    #
    typeAllLayers, nParamAllLayers, \
      nParamConvolutionalLayers, nParamFullyConnectedLayers, nParamBatchNormLayers = parseSummary(capturedSummary)
    #
    ### Print statistics
    nParamTot = sum(nParamAllLayers)
    print()
    print('Total number of parameters: {:9d}'.format(nParamTot))
    print()
    printStats('Convolutional     ',nParamConvolutionalLayers,nParamTot)
    printStats('FullyConnected    ',nParamFullyConnectedLayers,nParamTot)
    printStats('BatchNormalization',nParamBatchNormLayers,nParamTot)
    #
    ### Plot
    #
    plotParamsSummary(typeAllLayers, nParamAllLayers, minPctPlot)
    #
```

```python
### Interpretation of torchsummary.summary

#
## Conv2d-1 is a "Conv2d(1, 6, kernel_size=(5, 5), stride=(1, 1), padding=(2, 2))" layer that has
#
#    - Input: tensor of 28 x 28 x 1 (Width x Height x NChannels)
#
#    - 6 kernels of size 5x5 and number of channels=1 (the first parameter of summary)
#        => 150 = 6 * 5 * 5 * 1 two-body weights 
#             6 = 6 * 1         biases
#           156 TOTAL
#
#    - Output: 28 x 28 x 6
##
## Tanh-2 is a non-linearity layer without parameters
#
#    - Input:  tensor of 28 x 28 x 6
#    - Output: tensor of 28 x 28 x 6
#
## AvgPool2d-3 is an average pooling layer "AvgPool2d(kernel_size=2, stride=2, padding=0)"
#
#    - Input:  tensor of 28 x 28 x 6
#    - Output: tensor of 14 x 14 x 6
#
### Conv2d-4 is a "Conv2d(6, 16, kernel_size=(5, 5), stride=(1, 1))" layer that has
#
#    - Input: tensor of 14 x 14 x 6
#
#    - 16 kernels of size 5x5 and number of channels=6 (the number of channels of the input tensor)
#        => 2400 = 16 * 5 * 5 * 6 two-body weights 
#             16 = 16 * 1         biases
#           2416 TOTAL
#
#    - Output: 10 x 10 x 16  (padding = 'same')
#
### ...
#
### Linear-8 is a "Inner product" laver "Linear(in_features=400, out_features=120, bias=True)"
#
#    - Input: tensor of 400 x 1
#
#    - 120 hidden units
#        => 48000 = 120 * 400 two-nody weights
#             120 = 120 * 1   biases
#           48120 TOTAL
#
#    - Input: tensor of 120 x 1
#
### ...
#
```

```python
mLeNet5.parameters()
```

```python
nParamsLeNet5      = sum(p.numel() for p in mLeNet5.parameters())
nParamsLayerLeNet5 = [p.numel()    for p in mLeNet5.parameters()]
```

```python
print(nParamsLeNet5)
```

```python
print(nParamsLayerLeNet5)   ### two-body weights and biases at different indexes
```

```python
### Now print and plot the same things with a different format
parseSummaryAndPlot(summaryLeNet5)
```
